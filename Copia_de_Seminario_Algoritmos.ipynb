Sin restricciones, las formas de repartir las 30 tomas en días corresponden al número de Bell: B₃₀ = 846.749.014.511.809.332.450.147 ≈ 8,47·10²³.

Con las restricciones (máximo 6 tomas por día, sin fijar el número de días), se suman todas las combinaciones de días de 1 a 6 tomas que cubren las 30: 726.391.948.970.868.949.621.309 ≈ 7,26·10²³. La restricción reduce el espacio solo un 14 %, así que recorrerlo entero por fuerza bruta sigue siendo inviable.

En ambos casos se asume que el orden de las tomas dentro de un día y el orden de los días no afectan al coste: solo importa qué tomas se graban juntas.


```
#Imports
from math import factorial, comb
from itertools import product
```

```
#Respuesta
N_TOMAS = 30
MAX_TOMAS_DIA = 6

# SIN RESTRICCIONES: número de Bell B(30)
# stirling2(n, k) = formas de repartir n tomas en exactamente k días;
# sumando de 1 a 30 días se obtienen todas las agrupaciones posibles.
def stirling2(n, k):
    return sum((-1) ** j * comb(k, j) * (k - j) ** n for j in range(k + 1)) // factorial(k)

sin_restricciones = sum(stirling2(N_TOMAS, k) for k in range(1, N_TOMAS + 1))

# --- CON RESTRICCIONES ---
# m2..m6 = cuántos días de 2..6 tomas; m1 = tomas que sobran (días de 1 toma).
# Para cada combinación: 30! / producto(i!^mi · mi!), quitando órdenes repetidos.
con_restricciones = 0
for m in product(*(range(N_TOMAS // i + 1) for i in range(2, MAX_TOMAS_DIA + 1))):
    m1 = N_TOMAS - sum(i * mi for i, mi in zip(range(2, MAX_TOMAS_DIA + 1), m))
    if m1 >= 0:
        repetidas = factorial(m1)
        for i, mi in zip(range(2, MAX_TOMAS_DIA + 1), m):
            repetidas *= factorial(i) ** mi * factorial(mi)
        con_restricciones += factorial(N_TOMAS) // repetidas

print(f"Posibilidades Sin restricciones (Bell B30)     : {sin_restricciones:,}  ≈ {sin_restricciones:.2e}")
print(f"Posibilidades con restricciones (máx. 6 / día) : {con_restricciones:,}  ≈ {con_restricciones:.2e}")
```

Respuesta

Función objetivo. El coste de una planificación es la sumatoria, día por día, del número de actores distintos que deben ir al estudio ese día.
Como todos los actores cobran lo mismo por cada día que se desplazan, minimizar este número equivale a minimizar el gasto. Para saber qué actores van un día, se juntan los actores de todas las tomas que se graban ese día, contando una sola vez a cada actor aunque participe en varias de esas tomas.

Por ejemplo, si un día se graban las tomas 1 (actores 1, 2, 3, 4 y 5) y 2 (actores 3, 4 y 5), ese día deben ir 5 actores, no 8, porque los actores 3, 4 y 5 se repiten.

La solución debe cumplir las restricciones: cada toma se graba exactamente una vez y ningún día tiene más de 6 tomas. Si no las cumple, se considera no válida (coste infinito).

Es un problema de minimización, porque el objetivo es que el gasto por los servicios de los actores sea el menor posible.

En la práctica, grabar las tomas en orden de 6 en 6 cuesta 38 desplazamientos, mientras que agrupando tomas con actores parecidos el coste baja a 27.



```
# Restricciones: cada toma se graba una sola vez y máximo 6 tomas por día
def es_valida(solucion):
    todas = [t for dia in solucion for t in dia]
    return sorted(todas) == list(range(1, N_TOMAS + 1)) and all(len(dia) <= MAX_TOMAS_DIA for dia in solucion)

# Actores distintos que deben ir un día
def coste_dia(dia, datos=ACTORES_POR_TOMA):
    return len(set().union(*(datos[t - 1] for t in dia)))

# Función objetivo: sumatoria del coste de cada día
def funcion_objetivo(solucion):
    if not es_valida(solucion):
        return float("inf")
    return sum(coste_dia(dia) for dia in solucion)

# Comprobación con tres planificaciones
solucion_agrupada = [[1, 10, 11, 12, 22, 26], [14, 17, 18, 19, 23, 24], [3, 4, 8, 15, 21, 29],
                     [2, 13, 20, 27, 28, 30], [5, 6, 7, 9, 16, 25]]
solucion_invalida = [[1, 2, 3, 4, 5, 6, 7], list(range(8, 31))]

print("Tomas en orden :", funcion_objetivo(solucion_ejemplo))
print("Tomas agrupadas:", funcion_objetivo(solucion_agrupada))
print("Inválida       :", funcion_objetivo(solucion_invalida))
```

Diseña un algoritmo para resolver el problema por fuerza bruta

Respuesta

La fuerza bruta consiste en evaluar todas las planificaciones posibles, calcular el coste de cada una y seleccionar la que tenga el menor coste, sin aplicar atajos para reducir el número de opciones.

Para generar las planificaciones sin repetirlas, se utiliza una estrategia en la que la primera toma pendiente inicia un nuevo día. A partir de ahí, se prueban las diferentes combinaciones posibles con las demás tomas, teniendo en cuenta que no se pueden programar más de 6 tomas por día. Este proceso se repite con las tomas restantes hasta completar la planificación. De esta manera, se generan todas las posibilidades sin duplicarlas y se respeta la restricción desde el inicio.

Al aplicar este método a las 30 tomas, el número de planificaciones que se deben evaluar es demasiado elevado, aproximadamente 7,26 × 10²³, por lo que su ejecución resulta inviable. Por este motivo, se realizaron pruebas con las primeras 4, 6, 8, 10 y 12 tomas.El número de planificaciones obtenido coincide con el que dan las fórmulas del primer punto aplicadas a 4, 6, 8, 10 y 12 tomas, lo que permite comprobar que el algoritmo está generando las posibilidades correctamente.

Los resultados muestran que el número de planificaciones crece muy rápidamente a medida que aumenta la cantidad de tomas. Al pasar de 8 a 10 tomas, el número de posibilidades se multiplica casi por 28, y al pasar de 10 a 12 tomas, por más de 36 (de 115.274 a 4.163.743). El tiempo de ejecución crece en la misma proporción: de menos de un segundo con 10 tomas a mas de 40 segundos con 12.

```

from itertools import combinations
import time

# Genera todas las planificaciones válidas, sin repetir.
def generar_planificaciones(tomas):
    if not tomas:
        yield []
        return

    # La primera toma pendiente abre un día; se prueban todas sus posibles compañeras (máximo 5)
    primera, resto = tomas[0], tomas[1:]
    for n_companeras in range(min(MAX_TOMAS_DIA, len(tomas))):
        for companeras in combinations(resto, n_companeras):
            dia = [primera, *companeras]
            quedan = [t for t in resto if t not in companeras]
            for otros_dias in generar_planificaciones(quedan):
                yield [dia] + otros_dias

# Fuerza bruta: evalúa todas las planificaciones y se queda con la de menor coste
def fuerza_bruta(tomas):
    mejor_coste, mejor_solucion, revisadas = float("inf"), None, 0
    for solucion in generar_planificaciones(tomas):
        revisadas += 1
        coste = sum(coste_dia(dia) for dia in solucion)
        if coste < mejor_coste:
            mejor_coste, mejor_solucion = coste, solucion
    return mejor_coste, mejor_solucion, revisadas

# Se aplican a las primeras n tomas, con las 30 es imposible en este ejercicio por su larga duración
for n in (4, 6, 8, 10, 12):
    t0 = time.perf_counter()
    coste, solucion, revisadas = fuerza_bruta(list(range(1, n + 1)))
    print(f"{n:2d} tomas: {revisadas:>7,} planificaciones revisadas en {time.perf_counter() - t0:.2f} s "
          f"-> coste mínimo {coste}, días {solucion}")
```

Evaluar una planificación es rápido, porque solo hay que recorrer las n tomas y contar los actores de cada día, así que su coste es O(n). Lo que realmente pesa es la cantidad de planificaciones, que crece como el número de Bell, B(n). Por eso, la complejidad del algoritmo de fuerza bruta es O(n · B(n)), que crece incluso más rápido que una función exponencial.

(*)Diseña un algoritmo que mejore la complejidad del algortimo por fuerza bruta. Argumenta porque crees que mejora el algoritmo por fuerza bruta

Respuesta

Para mejorar la fuerza bruta se utiliza un algoritmo voraz. En lugar de revisar todas las planificaciones posibles, el algoritmo construye una sola, armando los días uno a uno: cada día empieza con la toma pendiente que tiene más actores, y se va completando con la toma que obliga a traer menos actores nuevos, hasta llegar a 6 tomas.

El algoritmo mejora la fuerza bruta porque no recorre todas las combinaciones, sino que en cada paso toma la decisión que parece mejor en ese momento. Así, resuelve las 30 tomas en menos de un segundo, mientras que la fuerza bruta tardaría miles de millones de años.

A cambio, no garantiza encontrar la mejor solución, ya que nunca vuelve atrás para corregir una decisión. Con 8 y 10 tomas obtiene el mismo coste que la fuerza bruta, pero con 12 tomas obtiene 15 frente a 14.

```

import time

# Algoritmo voraz: arma los días uno a uno.
# Cada día empieza con la toma que tiene más actores y se completa con la que
# obliga a traer menos actores nuevos (si empatan, la que comparte más actores).
def voraz(tomas, datos=ACTORES_POR_TOMA):
    pendientes = list(tomas)
    solucion = []
    while pendientes:
        inicial = max(pendientes, key=lambda t: len(datos[t - 1]))
        dia, actores_dia = [inicial], set(datos[inicial - 1])
        pendientes.remove(inicial)
        while len(dia) < MAX_TOMAS_DIA and pendientes:
            siguiente = min(pendientes, key=lambda t: (len(datos[t - 1] - actores_dia),
                                                      -len(datos[t - 1] & actores_dia)))
            dia.append(siguiente)
            actores_dia |= datos[siguiente - 1]
            pendientes.remove(siguiente)
        solucion.append(sorted(dia))
    return solucion

# Problema completo: 30 tomas
t0 = time.perf_counter()
solucion_voraz = voraz(range(1, N_TOMAS + 1))
print(f"30 tomas: coste {funcion_objetivo(solucion_voraz)} en {time.perf_counter() - t0:.4f} s")
for numero, dia in enumerate(solucion_voraz, start=1):
    print(f"  Día {numero}: tomas {dia} -> {coste_dia(dia)} actores")

# Comparación con la fuerza bruta en los mismos tamaños
for n in (8, 10, 12):
    print(f"{n} tomas: coste voraz {sum(coste_dia(d) for d in voraz(range(1, n + 1)))}")
```

(*)Calcula la complejidad del algoritmo

Respuesta

El algoritmo voraz coloca cada toma una sola vez, así que realiza n elecciones, una por cada toma. En cada elección revisa las tomas que quedan pendientes (como mucho n) y, para cada una, compara sus actores con los que ya van ese día, lo que cuesta O(A), siendo A el número de actores.

Por lo tanto, la complejidad del algoritmo es O(n · n · A) = O(n² · A). Como el número de actores es fijo, se puede expresar como O(n²), es decir, una complejidad cuadrática.

Esto supone una gran mejora frente a la fuerza bruta, cuya complejidad es O(n · B(n)), de tipo superexponencial.

Respuesta

Para generar datos aleatorios que tengan sentido, se tomó como referencia la forma de los datos reales: unos pocos actores (los protagonistas) participan en muchas tomas, mientras que la mayoría participa en pocas. Además, cada toma tiene al menos 2 actores, ya que una escena de doblaje normalmente implica un diálogo.

El generador recibe el número de tomas y de actores, y decide para cada toma qué actores participan: los protagonistas con una probabilidad del 60 % y los secundarios con una del 15 %. Se utiliza una semilla para que los datos se puedan reproducir exactamente. Los datos se generan como una lista de conjuntos, la misma estructura utilizada para el problema original.

Como ejemplo, se generó un juego de datos de 60 tomas y 15 actores, el doble del tamaño del problema original.

```
import random

# Genera datos aleatorios
def generar_datos(n_tomas, n_actores, n_protagonistas=3, prob_protagonista=0.6,
                  prob_secundario=0.15, semilla=None):
    rng = random.Random(semilla)
    datos = []
    for _ in range(n_tomas):
        actores = {a for a in range(1, n_actores + 1)
                   if rng.random() < (prob_protagonista if a <= n_protagonistas else prob_secundario)}
        while len(actores) < 2:
            actores.add(rng.randint(1, n_actores))
        datos.append(actores)
    return datos

datos_aleatorios = generar_datos(n_tomas=60, n_actores=15, semilla=42)

print("Primeras 5 tomas:")
for t, actores in enumerate(datos_aleatorios[:5], start=1):
    print(f"  Toma {t}: actores {sorted(actores)}")
print("Tomas por actor:", [sum(a in toma for toma in datos_aleatorios) for a in range(1, 16)])
```
